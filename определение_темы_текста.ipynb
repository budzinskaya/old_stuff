```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string

# загрузка необходимых данных NLTK
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
# функция для лемматизации и удаления стоп-слов
def preprocess_text(text):
    lemmatizer = WordNetLemmatizer()
    stop_words = set(stopwords.words('english'))

    # токенизация
    words = nltk.word_tokenize(text)

    lemmatized_words = list()
    for word in words:
     if word.lower() not in stop_words and word not in string.punctuation:
      lemmatized_words.append(lemmatizer.lemmatize(word.lower()))

    return lemmatized_words
```

```python
with open('Car history.txt', 'r') as cars, open('Caring for dogs.txt', 'r') as dogs, open('Origami.txt', 'r') as origami, open('Phonetics.txt', 'r') as phonetics, open('Roman Empire.txt', 'r') as roman:
  cars_lemmas = preprocess_text(cars.read()) #я всё препроцешшу сразу же, потому что не придумала как это по-умному сделать
  dogs_lemmas = preprocess_text(dogs.read())
  origami_lemmas = preprocess_text(origami.read())
  phonetics_lemmas = preprocess_text(phonetics.read())
  roman_lemmas = preprocess_text(roman.read())
```

```python
themes = {
    "Car history": cars_lemmas,
    "Dogs": dogs_lemmas,
    "Origami" : origami_lemmas,
    "Phonetics": phonetics_lemmas,
    "Roman Empire": roman_lemmas
}
```

```python
#загрузим и обработаем новый текст: Eminem and the “gay lisp”.txt
new_file = input("Введите название файла:")
with open(new_file, "r") as new:
  new_lemmas = preprocess_text(new.read()) # её мы тоже сразу обработаем, а почему нет
```

```python
max = 0
best_guess = ""
for key, value in themes.items():
  common = len(set(new_lemmas) & set(value))
  if common > max:
    max = common
    best_guess = key

print(best_guess)
```
